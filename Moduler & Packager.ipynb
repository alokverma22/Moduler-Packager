```python
import datetime
import time
import math
import random
import uuid
import os


# DATETIME AND TIME OPERATIONS
def datetime_menu():
    while True:
        print("\n================================")
        print("Datetime and Time Operations:")
        print("================================")
        print("1. Display current date and time")
        print("2. Calculate difference between two dates/times")
        print("3. Format date into custom format")
        print("4. Stopwatch")
        print("5. Countdown Timer")
        print("6. Back to Main Menu")
        print("================================")

        choice = input("Enter your choice: ")

        if choice == "1":
            now = datetime.datetime.now()
            print("Current Date and Time:",
                  now.strftime("%Y-%m-%d %H:%M:%S"))

        elif choice == "2":
            try:
                d1 = input("Enter the first date (YYYY-MM-DD): ")
                d2 = input("Enter the second date (YYYY-MM-DD): ")

                date1 = datetime.datetime.strptime(d1, "%Y-%m-%d")
                date2 = datetime.datetime.strptime(d2, "%Y-%m-%d")

                diff = abs(date2 - date1)
                print("Difference:", diff.days, "days")

            except ValueError:
                print("Invalid date format! Use YYYY-MM-DD.")

        elif choice == "3":
            now = datetime.datetime.now()
            fmt = input(
                "Enter custom format (e.g., %d-%m-%Y %H:%M:%S): "
            )

            try:
                print("Formatted Date:", now.strftime(fmt))
            except Exception:
                print("Invalid format string.")

        elif choice == "4":
            input("Press Enter to start stopwatch...")
            start = time.time()

            input("Press Enter to stop stopwatch...")
            end = time.time()

            print("Elapsed Time:", round(end - start, 2), "seconds")

        elif choice == "5":
            try:
                sec = int(input("Enter countdown time in seconds: "))

                while sec > 0:
                    print(sec, end="\r")
                    time.sleep(1)
                    sec -= 1

                print("Time's up!")

            except ValueError:
                print("Please enter a valid number.")

        elif choice == "6":
            break

        else:
            print("Invalid choice!")


# MATHEMATICAL OPERATIONS
def math_menu():
    while True:
        print("\n================================")
        print("Mathematical Operations:")
        print("================================")
        print("1. Calculate Factorial")
        print("2. Solve Compound Interest")
        print("3. Trigonometric Calculation")
        print("4. Area of Geometric Shapes")
        print("5. Back to Main Menu")
        print("================================")

        choice = input("Enter your choice: ")

        if choice == "1":
            try:
                n = int(input("Enter a number: "))

                if n < 0:
                    print("Please enter a positive integer.")
                else:
                    print("Factorial:", math.factorial(n))

            except ValueError:
                print("Please enter a valid integer.")

        elif choice == "2":
            try:
                p = float(input("Enter principal amount: "))
                r = float(input("Enter rate of interest (in %): "))
                t = float(input("Enter time (in years): "))

                amount = p * (1 + r / 100) ** t
                ci = amount - p

                print("Compound Interest:", round(ci, 2))

            except ValueError:
                print("Please enter valid numbers.")

        elif choice == "3":
            try:
                angle = float(input("Enter angle in degrees: "))
                rad = math.radians(angle)

                print("Sin:", round(math.sin(rad), 4))
                print("Cos:", round(math.cos(rad), 4))
                print("Tan:", round(math.tan(rad), 4))

            except ValueError:
                print("Please enter a valid angle.")

        elif choice == "4":
            try:
                r = float(input("Enter radius of circle: "))

                area = math.pi * r * r
                print("Area:", round(area, 2))

            except ValueError:
                print("Please enter a valid radius.")

        elif choice == "5":
            break

        else:
            print("Invalid choice!")


# RANDOM DATA GENERATION
def random_menu():
    while True:
        print("\n================================")
        print("Random Data Generation:")
        print("================================")
        print("1. Generate Random Number")
        print("2. Generate Random List")
        print("3. Create Random Password")
        print("4. Generate Random OTP")
        print("5. Back to Main Menu")
        print("================================")

        choice = input("Enter your choice: ")

        if choice == "1":
            try:
                start = int(input("Enter start: "))
                end = int(input("Enter end: "))

                print("Random Number:", random.randint(start, end))

            except ValueError:
                print("Please enter valid numbers.")

        elif choice == "2":
            items = input(
                "Enter elements separated by space: "
            ).split()

            if items:
                print("Selected:", random.choice(items))
            else:
                print("List is empty!")

        elif choice == "3":
            try:
                length = int(input("Enter password length: "))

                chars = (
                    "abcdefghijklmnopqrstuvwxyz"
                    "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
                    "0123456789!@#$%^&*()"
                )

                password = "".join(
                    random.choice(chars) for _ in range(length)
                )

                print("Generated Password:", password)

            except ValueError:
                print("Please enter a valid length.")

        elif choice == "4":
            otp = random.randint(100000, 999999)
            print("Generated OTP:", otp)

        elif choice == "5":
            break

        else:
            print("Invalid choice!")


# UUID GENERATION
def uuid_generate():
    print("\n================================")
    print("Generate Unique Identifiers (UUID)")
    print("================================")

    print("Generated UUID:", uuid.uuid4())


# FILE OPERATIONS
def file_menu():
    while True:
        print("\n================================")
        print("File Operations:")
        print("================================")
        print("1. Create a new file")
        print("2. Write to a file")
        print("3. Read from a file")
        print("4. Append to a file")
        print("5. Back to Main Menu")
        print("================================")

        choice = input("Enter your choice: ")

        if choice == "1":
            name = input("Enter file name: ")

            try:
                with open(name, "w") as file:
                    pass

                print("File created successfully!")

            except Exception as e:
                print("Error:", e)

        elif choice == "2":
            name = input("Enter file name: ")
            data = input("Enter data to write: ")

            try:
                with open(name, "w") as file:
                    file.write(data)

                print("Data written successfully!")

            except Exception as e:
                print("Error:", e)

        elif choice == "3":
            name = input("Enter file name: ")

            try:
                with open(name, "r") as file:
                    data = file.read()

                print("File content:")
                print(data)

            except FileNotFoundError:
                print("File not found!")

        elif choice == "4":
            name = input("Enter file name: ")
            data = input("Enter data to append: ")

            try:
                with open(name, "a") as file:
                    file.write(data)

                print("File appended successfully!")

            except Exception as e:
                print("Error:", e)

        elif choice == "5":
            break

        else:
            print("Invalid choice!")


# EXPLORE MODULE
def explore_module():
    mod_name = input(
        "\nEnter built-in module name (e.g., math, os, time): "
    )

    try:
        module = __import__(mod_name)

        print("\nAttributes in '" + mod_name + "':")
        print(dir(module))

    except ModuleNotFoundError:
        print("Module not found!")


# MAIN MENU
def main():
    while True:
        print("\n=======================================")
        print("Welcome to Multi-Utility Toolkit")
        print("=======================================")
        print("Choose an option:")
        print("1. Datetime and Time Operations")
        print("2. Mathematical Operations")
        print("3. Random Data Generation")
        print("4. Generate Unique Identifiers (UUID)")
        print("5. File Operations")
        print("6. Explore Module Attributes (dir())")
        print("7. Exit")
        print("=======================================")

        choice = input("Enter your choice: ")

        if choice == "1":
            datetime_menu()

        elif choice == "2":
            math_menu()

        elif choice == "3":
            random_menu()

        elif choice == "4":
            uuid_generate()

        elif choice == "5":
            file_menu()

        elif choice == "6":
            explore_module()

        elif choice == "7":
            print("\n============================")
            print("Thank you for using the Multi-Utility Toolkit!")
            print("============================")
            break

        else:
            print("Invalid choice!")


if __name__ == "__main__":
    main()
```
